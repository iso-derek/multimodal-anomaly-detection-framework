```python
import numpy as np  # generate or manipulate data
import pandas as pd  # work with CSV/table datasets
import matplotlib.pyplot as plt  # plot graphs

from sklearn.ensemble import IsolationForest  # Tree-based anomaly detector
from sklearn.neighbors import LocalOutlierFactor  # Local neighborhood anomaly detector
```

```python
## Create a Synthetic Dataset(proving our system works before using real data)

#2-feature synthetic dataset 
# We simulate normal behaviour around (50, 50).

# We inject fake anomalies around (100, 5).

# The model must detect these outliers.

rng = np.random.default_rng(42) # random number generator
# 200 data points [200 rows, 2 features per record] 
# data is centered around (50,50) 
#each feature varies with the standard deviation 8
normal_data = rng.normal(loc=[50, 50], scale=[8, 8], size=(200, 2)) 
outliers = rng.normal(loc=[100, 5], scale=[2, 2], size=(5, 2))

data = np.vstack([normal_data, outliers])
df = pd.DataFrame(data, columns=["feature_1", "feature_2"])
df.head()


```

```python
# Calculate the mean of each feature (average normal value)
means = df.mean()

# Calculate standard deviation of each feature (how spread out the values are)
stds = df.std()

# Compute the Z-score: how many standard deviations away each value is
z_scores = (df - means) / stds

# Flag row as anomaly if ANY feature has |Z| > 3 (3 = statistical threshold)
df["z_anomaly"] = (z_scores.abs() > 3).any(axis=1).astype(int)

# Count how many are anomalies (1) vs normal (0)
df["z_anomaly"].value_counts()
```

```python
### Isolation Forest + Local Outlier Factor

# Select the two features as the input matrix for the ML models
X = df[["feature_1", "feature_2"]].values


# -------------------- Isolation Forest --------------------
# 'contamination=0.03' tells the model we expect ~3% anomalies in the data
# 'random_state=42' ensures results are reproducible
iso = IsolationForest(contamination=0.03, random_state=42)

# Predict: IF returns 1 for normal and -1 for anomaly, so we convert -1 to 1
df["if_anomaly"] = (iso.fit_predict(X) == -1).astype(int)


# -------------------- Local Outlier Factor --------------------
# 'n_neighbors=20' compares each point with its 20 closest neighbors
# 'contamination=0.03' again specifies expected anomaly percentage
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.03)

# LOF also returns -1 for anomaly, so convert it to 1 like above
df["lof_anomaly"] = (lof.fit_predict(X) == -1).astype(int)


# Display first few anomaly results from the 3 detectors (Z, IF, LOF)
df[["z_anomaly", "if_anomaly", "lof_anomaly"]].head()
```

```python
# Ensemble = average of predictions
df["ensemble_score"] = (df["z_anomaly"] + df["if_anomaly"] + df["lof_anomaly"]) / 3
df["final_anomaly"] = (df["ensemble_score"] > 0.5).astype(int)

df["final_anomaly"].value_counts()
```

```python
##Display the results
plt.figure(figsize=(7,6))
plt.scatter(df["feature_1"], df["feature_2"], 
            c=df["final_anomaly"], cmap="coolwarm", s=60)
plt.title("Ensemble Anomaly Detection (Red = anomaly)")
plt.xlabel("feature_1")
plt.ylabel("feature_2")
plt.show()
```
