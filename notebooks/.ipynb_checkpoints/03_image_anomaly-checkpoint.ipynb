```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def load_image(path, size=(256, 256)):
    """
    Loads an image in RGB format and resizes it.
    Works for grayscale or colour images.
    """
    img = cv2.imread(path)

    if img is None:
        raise ValueError(f"Image at {path} could not be loaded.")

    # Convert BGR → RGB
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Resize to model input size
    img = cv2.resize(img, size)

    # Normalize to [0, 1]
    img = img.astype("float32") / 255.0

    return img


# Load your images (RGB)
normal_img = load_image("../images/normal_arm.jpg")
abnormal_img = load_image("../images/abnormal_arm.jpg")

plt.subplot(1,2,1); plt.title("Normal"); plt.imshow(normal_img); plt.axis("off")
plt.subplot(1,2,2); plt.title("Abnormal"); plt.imshow(abnormal_img); plt.axis("off")
plt.show()

```

```python
import torch

def to_tensor(img):
    """
    Converts an HWC image (RGB) to PyTorch CHW format.
    """
    tensor = torch.from_numpy(img).permute(2,0,1).unsqueeze(0)  # (1,3,H,W)
    return tensor.float()

normal_tensor = to_tensor(normal_img)
abnormal_tensor = to_tensor(abnormal_img)

normal_tensor.shape, abnormal_tensor.shape
```

```python
import torch.nn as nn

class ColorAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()

        # Encoder
        self.encoder = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, stride=2, padding=1),  # 3 → 32
            nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 128, kernel_size=3, stride=2, padding=1),
            nn.ReLU(),
        )

        # Decoder
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(128, 64, kernel_size=3, stride=2, padding=1, output_padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(64, 32, kernel_size=3, stride=2, padding=1, output_padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(32, 3, kernel_size=3, stride=2, padding=1, output_padding=1),
            nn.Sigmoid()  # output in range [0,1]
        )

    def forward(self, x):
        z = self.encoder(x)
        out = self.decoder(z)
        return out

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = ColorAutoencoder().to(device)
```

```python

# Helper function to load & resize images
def load_image(path):
    """
    Loads an image in ANY format:
    - If grayscale → convert to 3-channel RGB
    - If colour → leave as RGB
    Normalises and resizes.
    """
    img = cv2.imread(path, cv2.IMREAD_UNCHANGED)

    # If grayscale (2D array), convert to RGB
    if len(img.shape) == 2:
        img = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB)

    # If BGR colour, convert to RGB
    if img.shape[-1] == 3:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    img = cv2.resize(img, (256, 256))
    img = img.astype("float32") / 255.0
    return img


# Load normal and abnormal images
normal_img = load_image("../images/normal_arm.jpg")
abnormal_img = load_image("../images/abnormal_arm.jpg")

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(normal_img, cmap="gray"); plt.title("Normal X-ray")
plt.subplot(1, 2, 2); plt.imshow(abnormal_img, cmap="gray"); plt.title("Abnormal X-ray")
plt.show()
```

```python
# Convert RGB → Grayscale (needed for absdiff)
normal_gray = cv2.cvtColor(normal_img, cv2.COLOR_RGB2GRAY)
abnormal_gray = cv2.cvtColor(abnormal_img, cv2.COLOR_RGB2GRAY)

# Compute pixel-level absolute difference
diff = cv2.absdiff(normal_gray, abnormal_gray)

# Highlight large differences
_, thresh = cv2.threshold(diff, 30, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.title("Normal X-ray")
plt.imshow(normal_gray, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Abnormal X-ray")
plt.imshow(abnormal_gray, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("Difference Map (Anomalies Highlighted)")
plt.imshow(thresh, cmap="hot")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models


# Image Loading & Preprocessing

def load_image(path):
    """
    Loads an X-ray image in grayscale, resizes it to 256×256,
    and normalizes pixel values to the range [0,1].
    """
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)     # Load as grayscale
    img = cv2.resize(img, (256, 256))                # Resize for model
    img = img.astype("float32") / 255.0              # Normalize
    return img

# Load one normal and one abnormal X-ray image
normal_img = load_image("../images/normal_arm.jpg")
abnormal_img = load_image("../images/abnormal_arm.jpg")

# Autoencoder training data must ONLY contain normal images.
train_data = np.expand_dims(normal_img, axis=(0, -1))   # Shape: (1, 256,256,1)

# Format abnormal image for testing after training
abnormal_data = np.expand_dims(abnormal_img, axis=(0, -1))
```

```python
import torch.nn as nn

class ColorAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()

        # Encoder
        self.encoder = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, stride=2, padding=1),  # 3 → 32
            nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 128, kernel_size=3, stride=2, padding=1),
            nn.ReLU(),
        )

        # Decoder
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(128, 64, kernel_size=3, stride=2, padding=1, output_padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(64, 32, kernel_size=3, stride=2, padding=1, output_padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(32, 3, kernel_size=3, stride=2, padding=1, output_padding=1),
            nn.Sigmoid()  # output in range [0,1]
        )

    def forward(self, x):
        z = self.encoder(x)
        out = self.decoder(z)
        return out

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = ColorAutoencoder().to(device)
```

```python
with torch.no_grad():
    recon_normal = model(normal_tensor.to(device)).cpu().squeeze().permute(1,2,0).numpy()
    recon_abnormal = model(abnormal_tensor.to(device)).cpu().squeeze().permute(1,2,0).numpy()
```

```python
recon_normal.shape, recon_abnormal.shape
```

```python
import numpy as np

# generate 50 noisy versions of the normal image
train_imgs = []
for _ in range(50):
    noise = np.random.normal(0, 0.02, size=normal_img.shape).astype("float32")
    noisy = np.clip(normal_img + noise, 0, 1)
    train_imgs.append(noisy)

train_array = np.stack(train_imgs, axis=0)
train_tensor = torch.from_numpy(train_array).permute(0,3,1,2)  # (N,3,H,W)

train_dataset = torch.utils.data.TensorDataset(train_tensor, train_tensor)
train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=4, shuffle=True)
```

```python
import torch.optim as optim

criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=1e-3)

epochs = 8

model.train()
for e in range(epochs):
    epoch_loss = 0
    for batch, _ in train_loader:
        batch = batch.to(device)

        optimizer.zero_grad()
        recon = model(batch)
        loss = criterion(recon, batch)
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item()

    print(f"Epoch {e+1}/{epochs}, Loss = {epoch_loss/len(train_loader):.6f}")
```

```python
model.eval()
with torch.no_grad():
    normal_recon = model(normal_tensor.to(device)).cpu().squeeze().permute(1,2,0).numpy()
    abnormal_recon = model(abnormal_tensor.to(device)).cpu().squeeze().permute(1,2,0).numpy()

# squared error heatmap
normal_err = np.mean((normal_recon - normal_img)**2, axis=2)
abnormal_err = np.mean((abnormal_recon - abnormal_img)**2, axis=2)
```

```python
plt.figure(figsize=(14,8))

# Normal
plt.subplot(2,3,1); plt.title("Normal"); plt.imshow(normal_img); plt.axis("off")
plt.subplot(2,3,2); plt.title("Normal Recon"); plt.imshow(normal_recon); plt.axis("off")
plt.subplot(2,3,3); plt.title("Normal Error"); plt.imshow(normal_err, cmap="hot"); plt.axis("off")

# Abnormal
plt.subplot(2,3,4); plt.title("Abnormal"); plt.imshow(abnormal_img); plt.axis("off")
plt.subplot(2,3,5); plt.title("Abnormal Recon"); plt.imshow(abnormal_recon); plt.axis("off")
plt.subplot(2,3,6); plt.title("Abnormal Error"); plt.imshow(abnormal_err, cmap="hot"); plt.axis("off")

plt.show()
```

```python
from skimage.metrics import structural_similarity as ssim

# Convert images to grayscale
normal_gray = cv2.cvtColor(normal_img, cv2.COLOR_RGB2GRAY)
abnormal_gray = cv2.cvtColor(abnormal_img, cv2.COLOR_RGB2GRAY)

# Compute SSIM (structural similarity)
score, diff = ssim(
    normal_gray,
    abnormal_gray,
    full=True,
    data_range=1.0   # FIXED: since images are float in [0,1]
)

# Invert so anomalies (differences) are bright
diff = (1 - diff)

plt.figure(figsize=(12,6))

plt.subplot(1,3,1)
plt.title("Normal"); plt.imshow(normal_gray, cmap="gray"); plt.axis("off")

plt.subplot(1,3,2)
plt.title("Abnormal"); plt.imshow(abnormal_gray, cmap="gray"); plt.axis("off")

plt.subplot(1,3,3)
plt.title("SSIM Difference Map")
plt.imshow(diff, cmap="hot")
plt.axis("off")

plt.show()
```

Thresholding the Error Map into a Clean Binary Anomaly Mask

This takes your error map (either reconstruction error or SSIM anomaly map) and converts it into a:

Binary segmentation mask

0 = normal region

1 = anomaly region

This is exactly what medical imaging papers do when highlighting the fractures

```python
# Ensure anomaly map is grayscale (2D)
if len(anomaly_map_abnormal.shape) == 3:
    # Convert to grayscale by averaging channels
    anomaly_map_gray = anomaly_map_abnormal.mean(axis=2)
else:
    anomaly_map_gray = anomaly_map_abnormal

# Prepare for thresholding
anom_255 = (anomaly_map_gray * 255).astype("uint8")

# Otsu threshold
_, binary_mask = cv2.threshold(
    anom_255, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

binary_mask = binary_mask.astype("float32") / 255.0
```
